```python
df = (spark.read
      .option("header", "true")
      .option("inferSchema", "true")
      .csv("abfss://raw@hrdatask.dfs.core.windows.net/WA_Fn-UseC_-HR-Employee-Attrition.csv")
      )

df.write.mode("overwrite").format("delta").saveAsTable("workforce.bronze.employee_attrition_raw")
```

```python
from pyspark.sql.functions import col, sum as spark_sum, when 

null_summary = df.select([spark_sum(when(col(c).isNull(), 1).otherwise(0)).alias(c) for c in df.columns])

display(null_summary)
 
```
