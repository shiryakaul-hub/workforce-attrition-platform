```python
silver_df = (
    df
    .withColumnRenamed("EmployeeNumber","employee_id")
    .withColumnRenamed("Attrition","attrition")
    .withColumnRenamed("DailyRate","daily_rate")
    .withColumnRenamed("BusinessTravel","business_travel")
    .withColumnRenamed("Department","department")
    .withColumnRenamed("EmployeeCount","employee_count")
    .withColumnRenamed("DistanceFromHome","distance_from_home")
    .withColumnRenamed("Education","education")
    .withColumnRenamed("EducationField","education_field")
    .withColumnRenamed("EnvironmentSatisfaction","environment_satisfaction")
    .withColumnRenamed("Gender","gender")
    .withColumnRenamed("Age","age")
    .withColumnRenamed("HourlyRate","hourly_rate")
    .withColumnRenamed("JobInvolvement","job_involvement")
    .withColumnRenamed("JobLevel","job_level")
    .withColumnRenamed("JobRole","job_role")    
    .withColumnRenamed("JobSatisfaction","job_satisfaction")
    .withColumnRenamed("MaritalStatus","marital_status")
    .withColumnRenamed("MonthlyIncome","monthly_income")
    .withColumnRenamed("MonthlyRate","monthly_rate")
    .withColumnRenamed("NumCompaniesWorked","num_companies_worked")
    .withColumnRenamed("Over18","over_18")
    .withColumnRenamed("OverTime","over_time")
    .withColumnRenamed("PercentSalaryHike","percent_salary_hike")
    .withColumnRenamed("PerformanceRating","performance_rating")
    .withColumnRenamed("RelationshipSatisfaction","relationship_satisfaction")
    .withColumnRenamed("StandardHours","standard_hours")
    .withColumnRenamed("StockOptionLevel","stock_option_level")
    .withColumnRenamed("TotalWorkingYears","total_working_years")
    .withColumnRenamed("TrainingTimesLastYear","training_times_last_year")
    .withColumnRenamed("WorkLifeBalance","work_life_balance")
    .withColumnRenamed("YearsAtCompany","years_at_company")
    .withColumnRenamed("YearsInCurrentRole","years_in_current_role")
    .withColumnRenamed("YearsSinceLastPromotion","years_since_last_promotion")
    .withColumnRenamed("YearsWithCurrManager","years_with_curr_manager"

    )
)

display(silver_df)
```

```python
silver_df=(
    silver_df
.withColumn("age", col("age").cast("int"))
.withColumn("attrition", col("attrition").cast("boolean"))
.withColumn("business_travel", col("business_travel").cast("string"))
.withColumn("daily_rate", col("daily_rate").cast("int"))
.withColumn("department",col("department").cast("string"))
.withColumn("distance_from_home", col("distance_from_home").cast("int"))
.withColumn("education", col("education").cast("int"))
.withColumn("education_field", col("education_field").cast("string"))
.withColumn("employee_count", col("employee_count").cast("int"))

.withColumn("employee_id", col("employee_id").cast("int"))
.withColumn("environment_satisfaction", col("environment_satisfaction").cast("int"))
.withColumn("gender", col("gender").cast("string"))
.withColumn("hourly_rate", col("hourly_rate").cast("int"))
.withColumn("job_involvement", col("job_involvement").cast("int"))
.withColumn("job_level", col("job_level").cast("int"))
.withColumn("job_role", col("job_role").cast("string"))
.withColumn("job_satisfaction", col("job_satisfaction").cast("int"))
.withColumn("marital_status", col("marital_status").cast("string"))
.withColumn("monthly_income", col("monthly_income").cast("int"))
.withColumn("monthly_rate", col("monthly_rate").cast("int"))
.withColumn("num_companies_worked", col("num_companies_worked").cast("int"))
.withColumn("over_18", col("over_18").cast("boolean"))
.withColumn("over_time", col("over_time").cast("boolean"))
.withColumn("percent_salary_hike", col("percent_salary_hike").cast("int"))
.withColumn("performance_rating", col("performance_rating").cast("int"))
.withColumn("relationship_satisfaction", col("relationship_satisfaction").cast("int"))
.withColumn("standard_hours", col("standard_hours").cast("int"))
.withColumn("stock_option_level", col("stock_option_level").cast("int"))
.withColumn("total_working_years", col("total_working_years").cast("int"))
.withColumn("training_times_last_year", col("training_times_last_year").cast("int"))
.withColumn("work_life_balance", col("work_life_balance").cast("int"))
.withColumn("years_at_company", col("years_at_company").cast("int"))
.withColumn("years_in_current_role", col("years_in_current_role").cast("int"))
.withColumn("years_since_last_promotion", col("years_since_last_promotion").cast("int"))
.withColumn("years_with_curr_manager", col("years_with_curr_manager").cast("int"))
)
silver_df.printSchema()
```

```python
silver_df.select("monthly_income").describe().show()
silver_df.select("years_at_company").describe().show()
```

```python
silver_df=(

silver_df.withColumn(
    "tenure_band",

    when(col("years_at_company") <= 5, "0-5")
    .when((col("years_at_company") > 5) & (col("years_at_company") <= 10), "5-10")
    .when((col("years_at_company") > 10) & (col("years_at_company") <= 15), "10-15")
    .when((col("years_at_company") > 15) & (col("years_at_company") <= 20), "15-20")
    .when((col("years_at_company") > 20) & (col("years_at_company") <= 30), "20-30")
    .otherwise("30-40")
               
    )
)
```

```python
silver_df =(
silver_df.withColumn("salary_band",
                     when((col("monthly_income") > 0) & (col("monthly_income") < 5000), "Low")
                     .when((col("monthly_income") > 5000) & (col("monthly_income") <= 10000), "Medium")
                     .when((col("monthly_income") > 10000) & (col("monthly_income") <= 15000), "High")
                     .otherwise("Executive")
)
)    
```

```python
display(silver_df.limit(5))
```

```python
silver_df = (

    silver_df

    .withColumn(

        "attrition",

        when(
            col("attrition") == "true",
            1
        )

        .otherwise(0)

    )
 .withColumn(

        "over_18",

        when(
            col("over_18") == "true",
            1
        )

        .otherwise(0)

    )

    .withColumn(

        "over_time",

        when(
            col("over_time") == "true",
            1
        )

        .otherwise(0)

    )

)
```

```python
silver_df =(
     silver_df
    .withColumn(
        "silver_load_timestamp",
        current_timestamp()
    
))
```

```python
(silver_df
.write
.format("delta")
.mode("overwrite")
.saveAsTable("workforce.silver.employee")
)
```
